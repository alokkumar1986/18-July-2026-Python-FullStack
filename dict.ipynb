```python
#Write a program to swap the keys as values and values as keys from a dict
course = { 0: 'MySql', 1: 'Python', 2: 'Power BI'}

course1 = {}

for I, J in course.items():
  course1[J] = I

print(course1)

```

```python
course = { 0: 'MySql', 1: 'Python', 2: 'Power BI'}

course[3]= 'Django'

print(course)
```

```python
#Write a program to swap the keys as values and values as keys from a dict
course = { 0: 'MySql', 1: 'Python', 2: 'Power BI'}

course1 = { J:I for I, J in course.items()}  #shorthand form of for loop in dict

print(course1)
```
