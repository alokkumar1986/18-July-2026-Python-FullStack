```python
#Write a program to add two numbers
a = int(input("Enter the first number: "))
b = int(input("Enter the second number: "))
c = a + b
print("The sum is: ",c)
```

```python
#write a program to swap two numbers
a = int(input("Enter the first number: "))
b = int(input("Enter the second number: "))
c = a
a = b
b = c
print("The first number is: ",a)
print("The second number is: ",b)
```
