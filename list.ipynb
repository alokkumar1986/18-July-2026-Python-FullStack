```python
#consider the list of students
#s = ['sachin', 'sourav', 'dinesh', 'rahul', 'sehwag', 'dhoni', 'rohit', 'virat']
#Find the students whose starts with s.

s = ['sachin', 'sourav', 'dinesh', 'rahul', 'sehwag', 'dhoni', 'rohit', 'virat']
l = []

for i in s:
  if i.startswith('s'):
    l.append(i)

print(l)
```
