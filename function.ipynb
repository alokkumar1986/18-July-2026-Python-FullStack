```python
#write a program of calculator using function.

def findSum(a,b):
  return a+b

def findDiff(a,b):
  return a-b

def findMul(a,b):
  return a*b

def findDiv(a, b):
  return a/b

def findFloorDiv(a,b):
   return a//b

def findMod(a,b):
   return a%b

def findPower(a,b):
  return a**b

print('Sum : ', findSum(5,6))
print('Difference : ', findDiff(5,6))
print('Multiplication : ', findMul(5,6))
print('Division :', findDiv(5,6))
print('Floor Division :', findFloorDiv(5,6))
print('Modulus :', findMod(5,6))
print('Power :', findPower(5,6))


```
