```python
#write a program to check the number is palindrome

num = input('Enter a number : ')
if num[:] == num[::-1]:
   print('Palindrome')
else:
   print('Not palindrome')
```

```python
#write a program to input one digit number and print it in words

num = int(input('Enter a one digit number :'))

if num==0:
   print('Zero')
elif num==1:
   print('One')
elif num==2:
   print('Two')
elif num==3:
   print('Three')
elif num==4:
   print('Four')
elif num==5:
   print('Five')
elif num==6:
   print('Six')
elif num==7:
   print('Seven')
elif num==8:
   print('Eight')
elif num==9:
   print('Nine')
else:
  print('Wrong Number')
```
