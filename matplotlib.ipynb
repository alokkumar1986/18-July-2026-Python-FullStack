```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [2, 3, 4, 5]
plt.plot(x,y, 'red')
plt.title('X vs Y')
plt.xlabel('X')
plt.ylabel('Y')
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [10, 20, 30, 40, 50]
y = [62, 98, 170, 220, 310]

plt.plot(x,y, 'pink',linestyle='dashed',linewidth="2",  marker="*", markersize=5, markeredgecolor='blue')
plt.title('Runs scored by India \n India Vs SriLanka')
plt.xlabel('Overs')
plt.ylabel('Runs Scored')
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

x = ['0-10', '11-20', '21-30', '31-40', '41-50']
y = [62, 36, 72, 22, 90]

plt.bar(x,y, color="green", width=.9)
plt.title('Runs scored by India \n India Vs SriLanka')
plt.xlabel('Overs')
plt.ylabel('Runs Scored')
#plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

x = ['0-10', '11-20', '21-30', '31-40', '41-50']
y = [62, 36, 72, 22, 90]

plt.bar(x,y, color=["green", "blue", 'yellow', 'pink', 'brown'], width=[.1, .2, .3, .4, .9])
plt.title('Runs scored by India \n India Vs SriLanka')
plt.xlabel('Overs')
plt.ylabel('Runs Scored')
#plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

x = ['Male', 'Female', 'Transgender']
y = [ 40, 51, 9]

exp = [0, 0.2, 0.1]

plt.pie(y, labels=x, explode=exp, colors=['orange','green', 'red'], autopct='%.1f%%', startangle=180, shadow=True)
plt.legend()
plt.savefig('pie.pdf')
plt.show()
```
