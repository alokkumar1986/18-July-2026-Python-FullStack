```python
#write a program to print the numbers divisible by 2 between 1 and 100.

for item in range(1,101):
  if item % 2==0:
    print(item, end=" ")
```

```python
#write a program to print the numbers between 1 and 100 in descending order.

for i in range(100, 0, -1):
  print(i, end=" ")
```
